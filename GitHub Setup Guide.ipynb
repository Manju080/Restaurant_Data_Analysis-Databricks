````python
from databricks.sdk import WorkspaceClient
import os

# Initialize Workspace client (uses notebook authentication automatically)
w = WorkspaceClient()

# Define source and target
source_notebooks = [
    "/Users/manjutejas188@gmail.com/01_Bronze_layer",
    "/Users/manjutejas188@gmail.com/02_Silver_layer",
    "/Users/manjutejas188@gmail.com/03_Gold_layer",
    "/Users/manjutejas188@gmail.com/04_ML_Churn_Prediction"
]

target_dir = "/Users/manjutejas188@gmail.com/Restaurant_Data_Analysis-Databricks"

print("Copying notebooks using Databricks Workspace API...\n")
print(f"Target directory: {target_dir}\n")

for source_path in source_notebooks:
    notebook_name = os.path.basename(source_path)
    target_path = f"{target_dir}/{notebook_name}"
    
    try:
        # Export notebook content from source
        print(f"Reading: {source_path}")
        exported = w.workspace.export(path=source_path, format="SOURCE")
        
        # The export returns an ExportResponse object with .content attribute (base64)
        # Or it might directly return the content
        if hasattr(exported, 'content'):
            content_data = exported.content
        else:
            content_data = exported
        
        # Import to target location in Git repo
        print(f"Writing: {target_path}")
        # Check if target already exists
        try:
            existing = w.workspace.get_status(path=target_path)
            print(f"   ⚠️  Already exists: {target_path}")
            print(f"   Skipping to avoid overwrite.\n")
            continue
        except:
            # Target doesn't exist, safe to import
            pass
        
        w.workspace.import_(
            path=target_path,
            format="SOURCE",
            content=content_data,
            overwrite=False
        )
        
        print(f"✓ Successfully copied: {notebook_name}\n")
        
    except Exception as e:
        print(f"✗ Failed to copy {notebook_name}")
        print(f"   Error: {str(e)}\n")

print("="*70)
print("Notebooks copied! Now you can commit them to Git.")
print("="*70)
print("\nNext steps:")
print("1. Go to the Repos UI in Databricks")
print("2. You should see the 4 notebooks as 'Untracked' or 'Modified'")
print("3. Stage them, add a commit message, and push to GitHub")
````
